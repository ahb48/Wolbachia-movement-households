```python
%matplotlib inline
import numpy as np
import pylab as plt
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file into a DataFrame
ratios_df = pd.read_csv('ratios.csv')

# Convert the DataFrame to a dictionary for easier access
ratios_dict = {(eval(row['SexRatio']), row['ReleaseTimeStep']): row['Ratio'] for _, row in ratios_df.iterrows()}
ratios_dict
```

```python
# Load the CSV file into a DataFrame
wolb_av_df = pd.read_csv('wolb_av_all_hholds.csv')

# Convert the DataFrame to a dictionary for easier access
ratios_dict = {(eval(row['SexRatio']), row['ReleaseTimeStep']): row['WolbAv'] for _, row in wolb_av_df.iterrows()}
ratios_dict
```

```python
plt.figure(figsize=(7, 4))

bar_labels = [(1,1), (3,1), (3,3), (3,5), (3,10), (5,1), (5,3), (5,5), (5,10), (10,1), (10,3), (10,5), (10,10)]
width = 0.2

heights_10 = [
    ratios_dict[((1, 1), 10)],
    ratios_dict[((3, 1), 10)],
    ratios_dict[((3, 3), 10)],
    ratios_dict[((3, 5), 10)],
    ratios_dict[((3, 10), 10)],
    ratios_dict[((5, 1), 10)],
    ratios_dict[((5, 3), 10)],
    ratios_dict[((5, 5), 10)],
    ratios_dict[((5, 10), 10)],
    ratios_dict[((10, 1), 10)],
    ratios_dict[((10, 3), 10)],
    ratios_dict[((10, 5), 10)],
    ratios_dict[((10, 10), 10)]
]

heights_25 = [
    ratios_dict[((1, 1), 25)],
    ratios_dict[((3, 1), 25)],
    ratios_dict[((3, 3), 25)],
    ratios_dict[((3, 5), 25)],
    ratios_dict[((3, 10), 25)],
    ratios_dict[((5, 1), 25)],
    ratios_dict[((5, 3), 25)],
    ratios_dict[((5, 5), 25)],
    ratios_dict[((5, 10), 25)],
    ratios_dict[((10, 1), 25)],
    ratios_dict[((10, 3), 25)],
    ratios_dict[((10, 5), 25)],
    ratios_dict[((10, 10), 25)]
]

heights_50 = [
    ratios_dict[((1, 1), 50)],
    ratios_dict[((3, 1), 50)],
    ratios_dict[((3, 3), 50)],
    ratios_dict[((3, 5), 50)],
    ratios_dict[((3, 10), 50)],
    ratios_dict[((5, 1), 50)],
    ratios_dict[((5, 3), 50)],
    ratios_dict[((5, 5), 50)],
    ratios_dict[((5, 10), 50)],
    ratios_dict[((10, 1), 50)],
    ratios_dict[((10, 3), 50)],
    ratios_dict[((10, 5), 50)],
    ratios_dict[((10, 10), 50)]
]

heights_75 = [
    ratios_dict[((1, 1), 75)],
    ratios_dict[((3, 1), 75)],
    ratios_dict[((3, 3), 75)],
    ratios_dict[((3, 5), 75)],
    ratios_dict[((3, 10), 75)],
    ratios_dict[((5, 1), 75)],
    ratios_dict[((5, 3), 75)],
    ratios_dict[((5, 5), 75)],
    ratios_dict[((5, 10), 75)],
    ratios_dict[((10, 1), 75)],
    ratios_dict[((10, 3), 75)],
    ratios_dict[((10, 5), 75)],
    ratios_dict[((10, 10), 75)]    
]

heights_100 = [
    ratios_dict[((1, 1), 100)],
    ratios_dict[((3, 1), 100)],
    ratios_dict[((3, 3), 100)],
    ratios_dict[((3, 5), 100)],
    ratios_dict[((3, 10), 100)],
    ratios_dict[((5, 1), 100)],
    ratios_dict[((5, 3), 100)],
    ratios_dict[((5, 5), 100)],
    ratios_dict[((5, 10), 100)],
    ratios_dict[((10, 1), 100)],
    ratios_dict[((10, 3), 100)],
    ratios_dict[((10, 5), 100)],
    ratios_dict[((10, 10), 100)]    
]

positions = np.arange(len(bar_labels))
colors = ['C0', 'C1', 'C2', 'C3', 'C4']

positions_100 = positions - 2*width
positions_75 = positions - width
positions_50 = positions
positions_25 = positions + width
positions_10 = positions + 2*width

plt.style.use('ggplot')

# Plot bars side by side
plt.bar(positions_10, heights_10, width=width, color=colors[0], label=r'$\delta t = 10$')
plt.bar(positions_25, heights_25, width=width, color=colors[1], label=r'$\delta t = 25$')
plt.bar(positions_50, heights_50, width=width, color=colors[2], label=r'$\delta t = 50$')
plt.bar(positions_75, heights_75, width=width, color=colors[3], label=r'$\delta t = 75$')
plt.bar(positions_100, heights_100, width=width, color=colors[4], label=r'$\delta t = 100$')

# Set labels and title with font size 15
plt.xlabel('Sex Ratio (no. female, no. male)', color='black', fontsize=15)
plt.ylabel('Average Wolbachia-infected \n (per household)', color='black', fontsize=15)
plt.title('Non-releasing households', color='black', fontsize=15)

# Set tick labels with font size 15
plt.xticks(positions, labels=bar_labels, color='black', rotation=45, fontsize=15)
plt.yticks(color='black', fontsize=15)

# Set legend with font size 15
plt.legend(fontsize=13)

# Save and show the plot
plt.savefig('wolbachia_all_households_size.pdf', bbox_inches='tight')
plt.show()
```
